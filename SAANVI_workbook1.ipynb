```python
#Library, loops, and lists
#import two standard libraries
import numpy as np
import pandas as pd
#you can use "as" to nickname the libraries
```

```python
#inititalize an array
random_list = np.arange(stop = 100)
print(random_list)
```

```python
#A1 use np.array to create a new list, called even_list_1 that contains all the even numbers from 0 to 100 (exclusive)
#loop method
even_list_1 = np.array([i for i in random_list if i % 2 == 0]).tolist()
#step method
even_list_2 = np.arange(start=0, stop=100, step=2).tolist()
#both outputs
print(even_list_1)
print(even_list_2)
```

```python
#the [ ] allows us to access elements from an array
print(random_list[0]) #first element
#the : operator can be used to access elements across a certain range
print(random_list[5:15]) #elements 5-14
#we can also not use a number at either end of our : to indicate we want python to go all the way to the end of that direction
print(random_list[95:])
print(random_list[:5]) #first 5 elements
```

```python
#A2
print(random_list[90])
print(random_list[10:21])
print(random_list[90:])
```

```python
#doing mathmetical operations on lists
#want the list to start at 1 instead of 0
#add 1 to each element in the list
random_list = random_list + 1
print(random_list)
```

```python
#do the same with natural nlog, using np.log command
#take the natural log of each element in the array, and printing the result
print(np.log(random_list))
```

```python
#boolean indexing can also be performed
#boolean index array that shows where all the odd(falses) are
print(random_list % 2 == 0)
#taking in the boolean index array into the new array allows us to filter out the falses
even_list = random_list[random_list % 2 == 0]
print(even_list)
```

```python
#intitialize an empty array
even_list_3 = np.array([])

#for in in range just means for each value of i in the range of 0 to 100 (exclusive), do the following:
for i in range (0,101):
    #for each i value check if its even
    if i%2 == 0:
        #if it is even then append it to the arr=even_list_3 array and define values to be i
        even_list_3 = np.append(arr=even_list_3, values=i)
print(even_list_3)
```

```python
#make an array of numbers 0 to 200 (exlcusive)
big_list = np.arange(200)

#multiply numbers resulting in a list
even_list_4 = [2*x for x in big_list if x<= 50]

even_list_4 = np.array(even_list_4)
print(even_list_4)
```
